```python
import pandas as pd

df = pd.read_csv("data.csv")
df.head()
```

```python
df.info()
```

```python
df.columns
```

```python
df.info()

df = df.drop_duplicates()
df = df.dropna()
```

```python
import numpy as np

# Flight ID
df['Flight_ID'] = range(1, len(df)+1)

# Aircraft Type
df['Aircraft_Type'] = np.random.choice(['A320', 'B737', 'A321'], size=len(df))

# Capacity
capacity_map = {'A320':180, 'B737':160, 'A321':220}
df['Passenger_Capacity'] = df['Aircraft_Type'].map(capacity_map)

# Actual passengers
df['Actual_Passengers'] = (df['Passenger_Capacity'] * np.random.uniform(0.5, 0.95, len(df))).astype(int)

# Turnaround time
df['Turnaround_Time'] = np.random.randint(30, 90, len(df))
```

```python
print(df.columns)
```

```python
df['Route'] = df['Origin'] + "-" + df['Destination']
```

```python
def convert_to_minutes(x):
    x = str(x)
    hours = 0
    minutes = 0

    if 'h' in x:
        hours = int(x.split('h')[0])
        x = x.split('h')[1]

    if 'm' in x:
        minutes = int(x.split('m')[0].strip())

    return hours * 60 + minutes

df['Duration_Minutes'] = df['Duration Time'].apply(convert_to_minutes)
```

```python
df['Utilisation'] = df['Duration_Minutes'] / (df['Duration_Minutes'] + df['Turnaround_Time'])
```

```python
df[['Duration Time', 'Duration_Minutes']].head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.scatterplot(x='Duration_Minutes', y='Load_Factor', data=df)
plt.title("Duration vs Load Factor")
plt.show()
```

```python
sns.scatterplot(x='Duration_Minutes', y='Utilisation', data=df)
plt.title("Duration vs Utilisation")
plt.show()
```

```python
# Low occupancy routes
low_routes = df[df['Load_Factor'] < 0.6]
print(low_routes[['Route','Load_Factor']])

# Aircraft comparison
print(df.groupby('Aircraft_Type')['Load_Factor'].mean())

# Correlation
print(df[['Duration_Minutes','Load_Factor']].corr())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score

X = df[['Duration_Minutes', 'Turnaround_Time', 'Passenger_Capacity']]
y = df['Load_Factor']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("MAE:", mean_absolute_error(y_test, y_pred))
print("R2 Score:", r2_score(y_test, y_pred))
```
